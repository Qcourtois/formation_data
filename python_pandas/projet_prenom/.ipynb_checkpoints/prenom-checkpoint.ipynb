## 2 Quelle est la proportion totale Femmes / Hommes au cours du temps depuis 1900 ?
### Comment a-t-elle évolué au cours du temps ?

```python
df_an_sexe = df.groupby(["annais","sexe"])["nombre"].sum()

print(df_an_sexe)
```

```python
# Ici utilisation d'une boucle for pas optimal

serie_annee_prop = pd.Series()

for a in range(1900,2023):
    a = str(a)
    serie_annee_prop[a]=df_an_sexe[a][1]/(df_an_sexe[a][1]+df_an_sexe[a][2]) * 100

#Pourcentage d'Homme au fil du temps
print(serie_annee_prop)
serie_annee_prop.plot()
```

### [Plus dur] Combien de prénoms faut-il pour nommer 50% des bébés par année (e.g. en 1900 27 prénoms suffisent) ?

```python
s_a_p = df.groupby(["annais","preusuel"])["nombre"].sum()
print(s_a_p)
```

#### methode 1 ( mauvaise)

```python
#En 1900 METHODE 1 Avec un while et deux compteur
#J'extrait 1900 et je range par ordre décroissant
s_p = s_a_p["1900"].sort_values(ascending=False)

total = s_p.sum() #Total de naissance en 1900
#Je retire le prénom rares après avoir fait mon total
s_p = s_p.drop("_PRENOMS_RARES") 



somme_partiel = s_p.iloc[0]
compteur = 1
while somme_partiel < total/2:
    somme_partiel += s_p.iloc[compteur]
    compteur +=1

compteur
```

#### methode 2

```python
#En 1900 METHODE 2 avec cumsum
#J'extrait 1900 et je range par ordre décroissant
s_p = s_a_p["1900"].sort_values(ascending=False)

total = s_p.sum() #Total de naissance en 1900
#Je retire le prénom rares après avoir fait mon total
s_p = s_p.drop("_PRENOMS_RARES") 

s_p[s_p.cumsum()<total/2].size + 1  #+1 car on veut dépasser les 50% 
```

```python
s_nb_prenom_50 = pd.Series()
for a in range(1900,2023):
    a =str(a)
    s_p = s_a_p[a].sort_values(ascending=False)
    total = s_p.sum()
    s_p = s_p.drop("_PRENOMS_RARES")
    
    
    s_nb_prenom_50[a]=s_p[s_p.cumsum()<total/2].size + 1

s_nb_prenom_50.plot()
```

### La part des 10 prénoms les plus utilisés a-t-elle bien été divisée par 5 entre 1900 et aujourd'hui ?

```python
s_pars_10_p = pd.Series()
for a in range(1900,2023):
    a =str(a)
    s_p = s_a_p[a].sort_values(ascending=False)
    total = s_p.sum()
    s_p = s_p.drop("_PRENOMS_RARES")
    pars = (s_p[:10].sum()/total)*100
    s_pars_10_p[a]=pars

print(s_pars_10_p)
s_pars_10_p.plot()
```

La pars des 10 prénoms les plus donnés est effectivement passée de 33.1% a 6.7% ce qui représente bien environ un division par 5.

##  	5  [Plus dur] Qu'en pensez vous ? Extrait du Figaro :
De même, les Erwan, Enora (bretons), Maelys (occitan) courent dans toute la France, tout comme les Wilfried (anglo-saxon), Enzo, Nino (italiens), Medhi, Inès (arabes) ou Lola (espagnol) ont fait leur trou. À l'inverse, les Armel (breton), Maite (basque), Giulia (italien) restent majoritairement limités à une région.

```python

def region_prenom(df,prenom,seuil=1):
    return df.query("preusuel == @prenom and (nombre >= @seuil)").groupby('annais')["dpt"].count().drop('XXXX')

#region_prenom(df,"ERWAN",10)
```

```python
def nb_region_prenoms(df,list_prenom,seuil=1):
    df_n_r = pd.concat([region_prenom(df,nom,seuil) for nom in list_prenom],axis = 1,)
    df_n_r.columns = list_prenom
    df_n_r = df_n_r.sort_index()
    df_n_r.fillna(0, inplace=True) 
    return df_n_r
    
```

```python
def affiche_nb_region_prenoms(df,list_prenom,seuil = 1):
    ax = nb_region_prenoms(df,list_prenom,seuil).plot()
    s = str(seuil)
    ax.set_title("Nombre de departements où au moins "+ s + " enfant(s) ont été nommé(s) ainsi dans l'année")
    ax.set_xlabel('année')
    ax.set_ylabel("Nombre de dpt")
    plt.show()
    return None
```

```python
affiche_nb_region_prenoms(df,["ERWAN","ENORA","MAELYS"],5)
```

```python
affiche_nb_region_prenoms(df,["WILFRIED","ENZO","NINO","MEDHI","INÈS","LOLA"],1)
```

```python
affiche_nb_region_prenoms(df,["ARMEL","MAITE","GIULIA"],3)
```

### Reprenez votre dataframe des prénoms et essayez de rajouter une colonne 'origine' dans votre dataframe des prénoms. Essayez de le faire de deux manières : en utilisant un mapping avec dictionnaire (voir ressource) et avec la fonction merge de pandas.

```python
df_merge = df.merge(df_btn["origin"],left_on="preusuel",right_on="name",how='outer')
df_merge = df_merge.dropna(subset=["preusuel"])
df_merge
```

```python
d2 = dict(df_btn["origin"])
def origin(prenom):
    if prenom in d2:
        return d2[prenom]
    else:
        return float('nan')

df_merge_2 = pd.DataFrame(df)
df_merge_2["origin"]=df["preusuel"].map(origin)
df_merge_2
```
